```python
from pathlib import Path

source_path = Path("/Users/ishfaq/Downloads/Yelp JSON 2/yelp_dataset/yelp_academic_dataset_review.json")

save_dir = source_path.parent / "review_parts"
save_dir.mkdir(exist_ok=True)

parts = 10

# Count how many review records are in the file
with source_path.open("r", encoding="utf-8") as file:
    record_count = sum(1 for line in file)

base_size = record_count // parts
remaining = record_count % parts

print("Total records:", record_count)

with source_path.open("r", encoding="utf-8") as file:
    for part_number in range(1, parts + 1):
        rows_in_this_part = base_size

        if part_number <= remaining:
            rows_in_this_part += 1
            
# how many lines must be in current split file
        
        new_file = save_dir / f"reviews_part_{part_number}.json"

        with new_file.open("w", encoding="utf-8") as output:
            for _ in range(rows_in_this_part):
                output.write(file.readline())

        print(f"Created: {new_file}")

print("Done. The review file has been split successfully.")
```
